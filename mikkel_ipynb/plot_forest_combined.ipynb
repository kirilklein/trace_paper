```python
import sys
from pathlib import Path

# Add repo root to sys.path FIRST so local trace package takes precedence
repo_root = Path("/Users/rjm707/Desktop/trace_paper")
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

# Now import from local trace package
import polars as pl


cohort = "semaglutide-atc"
df = pl.read_csv(repo_root / f"data/{cohort}/pooled_results_log_rr.csv").sort("outcome").filter(
    pl.col("outcome") != "A10BJ"
).filter(
    ~pl.col("outcome").str.starts_with("V")
).filter(
    pl.col("method") == "IPW"
)
cohort = "diabetes_2"
df2 = pl.read_csv(repo_root / f"data/{cohort}/pooled_results_log_rr.csv").sort("outcome").filter(
    pl.col("outcome") != "A10BJ"
).filter(
    ~pl.col("outcome").str.starts_with("V")
).filter(
    pl.col("method") == "IPW"
)
cohort = "cvd"
df3 = pl.read_csv(repo_root / f"data/{cohort}/pooled_results_log_rr.csv").sort("outcome").filter(
    pl.col("outcome") != "A10BJ"
).filter(
    ~pl.col("outcome").str.starts_with("V")
).filter(
    pl.col("method") == "IPW"
)
```

```python
combined = df.join(df2, on="outcome", suffix="_diabetes").join(df3, on="outcome", suffix="_cvd")
```

```python
tab = pl.read_csv(repo_root / f"data/{cohort}_table.csv", columns=["outcome", "treated_prevalence"])
df = df.join(tab, on="outcome")
```

```python
atc_dict = pl.read_csv(repo_root / 'helper_data/atc_dict_en.csv', separator=";")
atc_dict = {atc[1:]: text.capitalize() for atc, text in zip(atc_dict['Code'], atc_dict['Text'])}
```

# Filtered Forest plot per Group

```python
# special_atc = [["A10"], ["C02", "C07", "C08", "C09"], ["G04BE"], ["N06A"], ["H03AA"], ["B01AF"], ["R0"], ["A02"], ["J01", "P01AB01"], ["N02BF"]]
special_atc = [["N06D"], ["M05B"], ["G04BE"], ["N06A"], ["A02"], ["N02BF"]]
# special_atc.pop(0)
special__atc = sum(special_atc, [])  # flatten list of lists
```

```python
filtered = combined.filter(
    pl.any_horizontal(
        pl.col("outcome").str.starts_with(atc) for atc in special__atc
    )
)
```

```python
N_group = []
for atc_group in special_atc:
    N = filtered.filter(
        pl.any_horizontal(
            pl.col("outcome").str.starts_with(atc) for atc in atc_group
        )
    ).height
    N_group.append(N)
```

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

import numpy as np

fig = plt.figure(figsize=(5, sum(N_group) * 0.5))
gs = gridspec.GridSpec(len(N_group), 1, height_ratios=N_group, figure=fig)

atc3_groups = ["N06D", "M05B", "G04B", "N06A", "A02", "N02B"]
atc3_groups = [atc_dict[g] for g in atc3_groups]

base_ax = fig.add_subplot(111, frameon=False)
base_ax.set_visible(False)
axes = [
    fig.add_subplot(gs[i], sharex=base_ax)
    for i in range(len(N_group))]

for i, ax in enumerate(axes):
    atc_group = special_atc[i]
    group_filtered = df.filter(
        pl.any_horizontal(
            pl.col("outcome").str.starts_with(atc) for atc in atc_group
        )
    )

    # Example data
    studies = group_filtered["outcome"]

    y = np.arange(0, len(studies)*2, 2)

    ax.errorbar(group_filtered["RR"], y, xerr=[group_filtered["RR"] - group_filtered["log_RR_CI95_lower"].exp(), group_filtered["log_RR_CI95_upper"].exp() - group_filtered["RR"]],
            fmt='o', color='black', ecolor='gray', capsize=3, markersize=3)
    ax.errorbar(group_filtered["RR_diabetes"], y-0.25, xerr=[group_filtered["RR_diabetes"] - group_filtered["log_RR_CI95_lower_diabetes"].exp(), group_filtered["log_RR_CI95_upper_diabetes"].exp() - group_filtered["RR_diabetes"]],
            fmt='o', color='red', ecolor='gray', capsize=3, markersize=3)
    ax.errorbar(group_filtered["RR_cvd"], y+0.25, xerr=[group_filtered["RR_cvd"] - group_filtered["log_RR_CI95_lower_cvd"].exp(), group_filtered["log_RR_CI95_upper_cvd"].exp() - group_filtered["RR_cvd"]],
            fmt='o', color='green', ecolor='gray', capsize=3, markersize=3)

    ax.axvline(1, color='red', linestyle='--', ymin=0, ymax=len(studies))  # reference line
    ax.set_yticks(y)
    ax.set_yticklabels(studies)
    ax.invert_yaxis()  # studies top-down
    ax.label_outer()               # hide x tick labels except on bottom axis
    ax.set_ylim(-0.5, len(studies)-0.5)
    ax.set_title(f"{atc3_groups[i]}")

from matplotlib.lines import Line2D
fig.legend(handles=[
    Line2D([0], [0], marker="o", color="red", label="Diabetes"),
    Line2D([0], [0], marker="o", color="black", label="Semaglutide"),
    Line2D([0], [0], marker="o", color="green", label="CVD"),
    ], loc="lower center", bbox_to_anchor=(0.5, -0.04), ncol=3)

fig.subplots_adjust(hspace=0.5)
ax.set_xlabel("Relative Risk (95% CI)")
fig.savefig(repo_root / "figures" / "forest_combined_atc.png", dpi=400, bbox_inches="tight")
fig.show()
```
